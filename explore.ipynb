```python
#Income by occupation
query = """
SELECT occupation,COUNT(*) as customer_count,ROUND(AVG(annual_income), 2) as avg_income,ROUND(MIN(annual_income), 2) as min_income,ROUND(MAX(annual_income), 2) as max_income
FROM customers
WHERE annual_income IS NOT NULL
GROUP BY occupation
ORDER BY avg_income DESC
LIMIT 10
"""
df = con.execute(query).df()
print(df)
```
